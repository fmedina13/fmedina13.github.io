Next, the assignment asks us to identify one numeric column in the dataset and find the mean, median, and mode of the data in that column. I chose the "Maximum flood depth" because I thought this would give the most insight into the severity of flooding events.

For the first part of the assignment, we are allowed to use functions from the pandas library to calculate this, so I did so below:

For the next part of the assignment, we were asked to find the mean, median, and mode of this column again, but only using base Python functions. This means that we cannot use the functions that pandas or other modules provide that make it easy to calculate these summary statistics.

To do this, I realized we needed to cycle through the values of the CSV, pull the ones that specifically corresponded to the "Max Flood" column, and store those values in a list so they could be used for analysis.

First, I opened the csv in read mode and stored the header line in a list. Then, I stored the index of the column header "Maximum Flood Depth (inches)" in a variable. Each following line in the CSV would store this value in the same position/index, so I could reuse this index to identify the necessary Max Flood Depth value in each row and store it in a list:

```python
# Opens the dataset, and reads the first line where the column headers live.

data = open("Street_Flooding_Events.csv", "r")

header = data.readline()

# Creates a list of the header values, and removes any unnecessary characters.
header_list = header.split(",")
header_list = [i.strip(' "\n') for i in header_list]

print(header_list)

# Identifies which index corresponds to the column for "Maximum Flood Depth".
# This allows us to index the other rows and pull the associated value for flood depth.

max_flood_index = header_list.index("Maximum Flood Depth (inches)")

print(max_flood_index)

# Below, I begin to put all the Maximum Flood Depth values into a list.

# Creates an empty list, and begins looping through each line of the dataset.
flood_depth_values = []

row = data.readline()

while row:
    # Within the loop, we create a list of the row values and clean the values.
    row_list = row.split(",")
    row_list = [i.strip(' "\n') for i in row_list]

    # Using the index gathered in the previous step, we identify the value in that row
    # associated with max flood depth and add it to the flood values list.

    flood_value = row_list[max_flood_index]
    flood_depth_values.append(flood_value)
    row = data.readline()


# Checks to make sure values were added and that the length of the list matches
# the number of rows in the original dataset (3611).
print(flood_depth_values)

len(flood_depth_values)
```

Now that the values have been stored in a list, I can use that list to calculate the mean, median, and mode of the data. 

To enable this analysis, I converted the values from a string to a float, and also created a new list that puts the values in order since this will be useful for calculating the median.

For mean and mode, I created a loop that cycles through all the values, counts how often those values occur, and adds those values (so that I can then divide by the total number of values to get the mean).

For median, I created a conditional statement that finds the median based on the number of values in the data.

```python
# Converts the flood depth values to floats in order to sort the values within a new list.

flood_depth_values = [float(i) for i in flood_depth_values]
ascending_flood_list = sorted(flood_depth_values)

# Using a loop for to set up calculations for mean and mode.


# Setting up a dictionary to keep count of the number of occurrences of each value.
mode_dict = {}

sum = 0

for i in ascending_flood_list:
    if i in mode_dict:
        mode_dict[i] += 1
    else:
        mode_dict[i] = 1

    # Also summing each value in the list, then dividing by list length to calculate mean.
    sum += i

mean = sum / len(ascending_flood_list)

# Looping through the keys in the mode dictionary, which are the values of max flood data.

count = 0

for key in mode_dict:
    # If the value associated with a key is higher than any previous value,
    # it will be assigned to the variable "count" and the key associated with that count
    # becomes the new mode.

    if mode_dict[key] > count:
        count = mode_dict[key]

        # By the end of the loop, the key that occurs most often in the data will be the mode.
        mode = key

# Since ascending_flood_list is in order, we can index it to find the median.

median = 0
num_values = len(ascending_flood_list)

# If the length of the list is odd, we can divide the length by 2 and add 1 to get the index.

if num_values % 2 != 0:
    median = ascending_flood_list[(num_values // 2) + 1]

# If the length is even, we must find the average of the middle two values.

else:
    median = (
        ascending_flood_list[num_values // 2]
        + ascending_flood_list[(num_values // 2) + 1]
    ) / 2

print("Mean:", mean)

print("Median:", median)

print("Mode:", mode)
```

Finally, we were asked to create a data visualization using the data. The data/calculations were allowed to come through pandas, but the drawing code could only use the Python standard library.

For my visualization, I decided to analyze the number of flooding events per borough whose maximum depth exceeded the mean maximum depth. In other words, I was trying to analyze which boroughs had higher frequencies of severe flooding events.

However, there was not yet a column in the dataset which tracked the borough of the flooding event. Instead, this information could be found within one of the other columns - "Sensor Name". But this column also includes other location data we don't care about. So we needed to isolate the borough in "Sensor Name" for each row, and then store that value in a new column labeled "Borough".

```python
# Creates a copy of the dataframe that we can edit without changing the original data.
# Since the analysis is about borough-level flood data, named it borough_flood_data

borough_flood_data = flood_data.copy()

# Creates a new column in the data frame that identifies the borough.

for index in borough_flood_data.index:
    # Looping through each row of the new data frame to identify which borough is associated with.
    # Since the values within the "Sensor Name" column include the borough, I am slicing the
    # first two letters of the "Sensor Name" in each row, and assigning that to a variable.

    sensor = borough_flood_data.loc[index, "Sensor Name"]
    borough = sensor[:2]

    # Renaming those letters to get a cleaner name for the boroughs.

    if borough == "BK":
        borough = "Brooklyn"
    elif borough == "BX":
        borough = "Bronx"
    elif borough == "Q ":
        borough = "Queens"
    elif borough == "M ":
        borough = "Manhattan"
    elif borough == "SI":
        borough = "Staten Island"

    # Finally, I assign the appropriate borough value to each row based on index.

    borough_flood_data.loc[index, "Borough"] = borough

# Pulled up the first 10 rows of this new dataframe to ensure the borough values populated correctly.
borough_flood_data.head(10)
```
