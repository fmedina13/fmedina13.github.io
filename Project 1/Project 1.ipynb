```python
import pandas as pd
```

```python
flood_data = pd.read_csv("Street_Flooding_Events.csv")
```

```python
flood_data.head()
```

```python
max_depth = flood_data["Maximum Flood Depth (inches)"]
```

```python
mean_depth = max_depth.mean()
median_depth = max_depth.median()
mode_depth = max_depth.mode()

print("Mean:", mean_depth)
print("Median:", median_depth)
print("Mode:", mode_depth)
```
